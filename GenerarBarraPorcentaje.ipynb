```python
# ✅ Instalar Pillow (normalmente ya está en Colab)
!pip install Pillow --quiet

from PIL import Image, ImageDraw

def generar_barra_porcentaje(porcentaje, ancho=800, alto=150,
                             color_base="#E0E0E0", color_progreso="#7CB342",
                             nombre_archivo="barra.png"):
    """
    Genera una imagen PNG con fondo transparente que representa una barra de progreso
    sin texto del porcentaje.
    """
    # Crear imagen transparente (RGBA)
    img = Image.new("RGBA", (ancho, alto), (0, 0, 0, 0))
    draw = ImageDraw.Draw(img)

    # Dibujar barra base (gris claro)
    draw.rectangle([0, alto/2 - 20, ancho, alto/2 + 20], fill=color_base)

    # Dibujar barra de progreso (verde)
    progreso = int(ancho * (porcentaje / 100))
    draw.rectangle([0, alto/2 - 20, progreso, alto/2 + 20], fill=color_progreso)

    # Guardar imagen como PNG con fondo transparente
    img.save(nombre_archivo, "PNG")
    print(f"✅ Imagen guardada correctamente: {nombre_archivo}")

# 🧩 Ejemplo de uso
generar_barra_porcentaje(78, nombre_archivo="barra_78.png")
```
